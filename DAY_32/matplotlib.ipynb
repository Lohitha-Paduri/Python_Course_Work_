```python
import matplotlib.pyplot as plt
hours = [1,2,3,4,5,6,7,8]
visitors = [30,25,60,65,70,59,79,80]
plt.plot(hours, visitors,
         color = "orange",
         marker = "s",
         linestyle = "--",
         linewidth = 2,
         alpha = 0.5,
         label = "Visitors")

plt.title("Website Visitors Per Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of visitors")
plt.legend()

plt.grid(True, linestyle="--",alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
brands = ["Apple", "Samsung", "Realmi", "Oppo", "Others"]
market_share = [30, 25, 20, 33, 15]
plt.pie(market_share,
        labels = brands,
        autopct = "%.2f%%",
        startangle = 90,
        colors = ["Skyblue", "gold", "lightcoral","violet","pink"],
        explode = [0, 0, 0.3,0,0],
        shadow = True
        )
plt.title("Mobile Market Share 2025")
plt.show()
```

```python
import matplotlib.pyplot as plt
ad_spend = [2,4,5,7,8,11,13,14]
sales = [27,25,30,15,45,54,65,32]
profit = [200,250,300,255,376,289,400,550]
colors = [9,1,2,5,6,3,7,8]

plt.scatter(ad_spend,sales,
            s = profit,
            c = colors,
            cmap = "Pastel1",
            alpha = 0.5,
            edgecolors = 'skyblue',
            linewidths = 1.2
            )
plt.title("Advertising Spend vs Sales(with Color & Size)", fontsize=14, fontweight = "bold")
plt.xlabel("Ad Spend($1000s)", fontsize = 12)
plt.ylabel("Sales ($1000s)", fontsize = 12)

plt.colorbar(label="Sales Value")

plt.grid(True, linestyle = "--",alpha = 0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
scores = [
    [70, 68, 86, 78, 89, 66, 69, 9],
    [65, 70, 68, 72, 75, 78, 79, 66],
    [89, 98, 87, 88, 83, 85, 84, 80],
    [60, 61, 62, 63, 64, 65, 66,90],
    [79, 86, 98, 67, 90, 56, 8, 78]
]
plt.boxplot(scores,
            patch_artist = True,
            medianprops = dict(color = "red", linewidth =1),
            whiskerprops = dict(color = "green",linewidth = 1),
            capprops = dict(color = "black",linewidth = 1),
            notch = False,
            vert = True,
            labels = ["A", "B", "C", "D", "E"]
            )
plt.title("Exam Score Distribution Across Classes")
plt.xlabel("Classes")
plt.ylabel("Scores")
plt.grid(True, linestyle= "--",alpha=0.7)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

data = np.array([
    [30,32,31,29,28,27,26],
    [33,34,32,30,29,28,27],
    [35,36,37,34,33,29,20],
    [37,38,39,32,28,29,37],
    [36,37,28,29,38,39,33]

])

days = ["Mon", "Tues", "Wed", "Thurs", "Fri", "Sat", "Sun"]
weeks = ["Week1", "Week2", "Week3", "Week4", "Week5"]

sns.heatmap(data,
            annot = True,
            fmt=".2f",
            cmap="YlOrRd",
            linewidth = 0.5,
            xticklabels = days,
            yticklabels = weeks

            )
plt.title("Weekly Temperature Heatmap", fontsize=14,fontweight = "bold")
plt.xlabel = ("Days of the week")
plt.ylabel = ("Weeks")
plt.show()


```
