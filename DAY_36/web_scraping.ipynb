```python
#Loading required libraries
import requests
from bs4 import BeautifulSoup
```

```python
URL = 'https://codegnan.com/'
```

```python
# Loading the webpage in memory using requests library
page = requests.get(URL)
```

```python
#check the status of the code page
page.status_code
```

```python
htmlCode = page.text
print(type(htmlCode))
```

```python
htmlCode = page.text
print(htmlCode)
```

```python
#BeautifulSoup organizes/parses that HTML so you can scrape information from it.
soup = BeautifulSoup(htmlCode, 'html.parser')
print(soup)
```

```python
title = soup.title.text
print("Website Title:", title)
```

```python
headings = soup.find_all("h1")
for heading in headings:
  print(heading.text.strip())
```

```python
links = soup.find_all("a")
for link in links[:20]: #first 10 links
  print(link.get("href"))
```

```python
images = soup.find_all("img")
for image in images[:20]:
  print(image.get("src"))
```

```python
description = soup.find("meta",attrs = {"name":"description"})
if description:
  print("Description:")
  print(description.get("content"))
```

```python
scripts = soup.find_all("script")
for script in scripts:
  print(script.get("src"))
```

```python
og_image = soup.find("meta", property = "og:image")
print(og_image["content"])
```

```python
og_title = soup.find("meta", property = "og:title")
print(og_title["content"])
```

```python
meta_tags = soup.find_all("meta")
for tag in meta_tags:
  print(tag.get("name"))
```
